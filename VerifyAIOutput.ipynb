```python
import pandas as pd

# Load datasets
df_2025 = pd.read_csv('2025.csv')
df_2026 = pd.read_csv('2026PartYear.csv')

# Filter for Total U.S.
us_2025 = df_2025[df_2025['Geography'] == 'Total U.S.']
us_2026 = df_2026[df_2026['Geography'] == 'Total U.S.']

# 1. Conventional Average Selling Price (ASP) Comparison
mean_2025_conv_asp = us_2025[us_2025['Type']=='Conventional']['ASP Current Year'].mean()
mean_2026_conv_asp = us_2026[us_2026['Type']=='Conventional']['ASP Current Year'].mean()
asp_change = (mean_2026_conv_asp - mean_2025_conv_asp) / mean_2025_conv_asp * 100

print(f"2025 Conventional Mean ASP: ${mean_2025_conv_asp:.2f}")
print(f"2026 YTD Conventional Mean ASP: ${mean_2026_conv_asp:.2f}")
print(f"ASP Change: {asp_change:.2f}%")

# 2. Conventional Total Volume Comparison (Full 2025 vs 2026 YTD available weeks)
mean_2025_conv_vol = us_2025[us_2025['Type']=='Conventional']['Total Bulk and Bags'].mean()
mean_2026_conv_vol = us_2026[us_2026['Type']=='Conventional']['Total Bulk and Bags'].mean()
vol_change = (mean_2026_conv_vol - mean_2025_conv_vol) / mean_2025_conv_vol * 100

print(f"\n2025 Conventional Mean Weekly Volume: {mean_2025_conv_vol:,.0f}")
print(f"2026 YTD Conventional Mean Weekly Volume: {mean_2026_conv_vol:,.0f}")
print(f"Volume Change: {vol_change:.2f}%")

# 3. Organic vs Conventional Price Spread (Organic ASP / Conventional ASP)
org_2025_asp = us_2025[us_2025['Type']=='Organic']['ASP Current Year'].mean()
org_2026_asp = us_2026[us_2026['Type']=='Organic']['ASP Current Year'].mean()
spread_2025 = (org_2025_asp - mean_2025_conv_asp) / mean_2025_conv_asp * 100
spread_2026 = (org_2026_asp - mean_2026_conv_asp) / mean_2026_conv_asp * 100

print(f"\n2025 Organic Price Premium over Conventional: {spread_2025:.1f}%")
print(f"2026 YTD Organic Price Premium over Conventional: {spread_2026:.1f}%")
```
